```python
#같은 변수명을 끌고 와 쓸거면 global 사용
#같은 변수명을 사용시 들여쓰기가 많은 위치에있는 변수부터 사용(자기안에있는 변수부터 사용한다)
def outer_func():
    global y #전역 변수불러오기
    y=20 #전역 변수 의 값 변경
    def inner_func():
        global y
        global z #밖에서 선언안했지만 밖에서 선언한 취급
        z = 10
        y=30
        print('inner y=',y)
    inner_func()
    print('outer y=',y)
y=10 #전역 변수
outer_func()
print('global y=',y)
print('z=',z)
```
