```python
import csv
import requests
from bs4 import BeautifulSoup
import os
import time
import random
```

```python
#접근 가능한 사이트인지 확인 필요
urls = ["https://www.imdb.com/title/tt7678620/episodes/?ref_=tt_ov_epl", 
       "https://moricoffee.co.kr/smoothie_freppe",
       "https://ko.wikipedia.org/wiki/파이썬"]
hdr = {'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/151.0.7922.174 Safari/537.36'}# 웹브라우저에 대한 정보

for url in urls:
       res = requests.get(url, headers=hdr)
       time.sleep(random.randint(3, 5)) # 작업대기 상태를 3초에서 5초 사이
       print("-" * 40)
       print(f"대상 URL: {url}, 상태코드: {res.status_code}")
       print(f"수집 가능: {'🟢 허용됨' if res.status_code == 200 else '🔴 차단됨 (수집 금지)'}")
```

```python
# 웹 페이지에서 코드 가져오기
url = "https://www.lego.com/ko-kr/themes/architecture"
# url = "https://www.lego.com/ko-kr/themes/technic"
hdr = {'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/151.0.7922.174 Safari/537.36'}
res = requests.get(url, headers=hdr)
time.sleep(3)

if res.status_code != 200:
    print(f"웹페이지응답코드:{res.status_code}, 웹 페이지 정보를 불러오지 못했습니다.")
else:
    res.raise_for_status() #코드를 가져오다가 문제가 생기면 예외처리해달라
    soup = BeautifulSoup(res.text, "lxml") # 페이지 자체에 어느정도가 있어도 스스로 보완해서 가져온다 (성능도 좋고 복원력도 좋음)
    #soup = BeautifulSoup(res.text, "html.parser")
    print(soup)
    
```

```python
#원하는 정보를 가지고 있는/획득할 수 있는 tag 찾기
products = soup.find_all('div', attrs={'class':'ListingProductLeaf_listingLeafWrapper__6kHno undefined'})
nums_products = len(products)
text_products = []
for s in products:
    print(s)
    text_products.append(s.get_text())

print(f"총 제품 수:{nums_products}")
print(f"포함된 텍스트:{text_products}")
```

```python
img_dir_name = "레고이미지"
if not os.path.isdir(img_dir_name):
      os.mkdir(img_dir_name)
cur_dir = os.getcwd()

for product in products:
    name = product.find('a', attrs={'data-test':'product-leaf-title'}).get_text() 
    age = product.find('span', attrs={'data-test':'product-leaf-age-range-desktop-label'}).get_text()
    num_piece = product.find('span', attrs={'data-test':'product-leaf-piece-count-desktop-label'}).get_text()
    price = product.find('span', attrs={'data-test':'product-leaf-price'}).get_text()
    dis_price = product.find('span', attrs={'data-test':'product-leaf-discounted-price'})
    if dis_price != None:
        dis_price = product.find('span', attrs={'data-test':'product-leaf-discounted-price'}).get_text()
    else: dis_price = 'NoSale'
    print(name, age, num_piece, price, dis_price)
    
    # img = product.find("picture", attrs={'data-test':'product-leaf-image-1'})
    imgurl = product.find('picture').find('source').get('srcset').split('?')[0]
    print(product)
    print(imgurl)
    r = requests.get(imgurl, headers=hdr)
    with open(f'{cur_dir}/{img_dir_name}/{name}.jpg', "wb") as out:
            out.write(r.content)
```

```python
filename = "레고_제품_정보.csv"
f = open(filename, "w", encoding="utf-8-sig", newline="")
writer = csv.writer(f)
attributes = ["제품명", "사용연령", "피스수", "정가", "할인가"]
writer.writerow(attributes)

url = "https://www.lego.com/ko-kr/themes/technic"
hdr = {'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/151.0.7922.174 Safari/537.36'}
res = requests.get(url, headers=hdr)
time.sleep(3)

if res.status_code != 200:
    print(f"웹페이지응답코드:{res.status_code}, 웹 페이지 정보를 불러오지 못했습니다.")
    exit()
else:
    res.raise_for_status()
    soup = BeautifulSoup(res.text, "lxml")
    #soup = BeautifulSoup(res.text, "html.parser")
    #print(soup)
    page_class = soup.find_all('div', attrs={'class':'ListingPagination_container__haX_Q ListingPagination_desktop__Tnoiu'})
    num_pages = page_class[0].find_all('li', attrs={'class':'r5YyRa_pagination-item default-focus-ring ds-label-md-medium'})
    print(len(num_pages))
    
if len(num_pages) == 1:
    products = soup.find_all('div', attrs={'class':'ListingProductLeaf_listingLeafWrapper__6kHno undefined'})
    
    for product in products:
        name = product.find('a', attrs={'data-test':'product-leaf-title'}).get_text() 
        age = product.find('span', attrs={'data-test':'product-leaf-age-range-desktop-label'}).get_text()
        num_piece = product.find('span', attrs={'data-test':'product-leaf-piece-count-desktop-label'}).get_text()
        price = product.find('span', attrs={'data-test':'product-leaf-price'}).get_text()
        dis_price = product.find('span', attrs={'data-test':'product-leaf-discounted-price'})
        if dis_price != None:
            dis_price = product.find('span', attrs={'data-test':'product-leaf-discounted-price'}).get_text()
        else: dis_price = 'NoSale'
        print(name, age, num_piece, price, dis_price)
        
        # img = product.find("picture", attrs={'data-test':'product-leaf-image-1'})
        imgurl = product.find('picture').find('source').get('srcset').split('?')[0]
        print(product)
        print(imgurl)
        r = requests.get(imgurl, headers=hdr)
        with open(f'{cur_dir}/{img_dir_name}/{name}.jpg', "wb") as out:
                out.write(r.content)
            
        prod_inf = [name, age, num_piece, price, dis_price]
        writer.writerow(prod_inf)
else:
    for page in range(len(num_pages)):
        time.sleep(3)
        url_page = url
        if page != 0:
            url_page += f'?page={page+1}&offset=1'
        print(url_page)
        res = requests.get(url_page, headers=hdr)        
        soup = BeautifulSoup(res.text, "lxml")
        
        
        products = soup.find_all('div', attrs={'class':'ListingProductLeaf_listingLeafWrapper__6kHno undefined'})
            
        for product in products:
            name = product.find('a', attrs={'data-test':'product-leaf-title'}).get_text() 
            
            if product.find('span', attrs={'data-test':'product-leaf-age-range-desktop-label'}):
                age = product.find('span', attrs={'data-test':'product-leaf-age-range-desktop-label'}).get_text()
            else:
                age = product.find('span', attrs={'data-test':'product-leaf-age-range-label'}).get_text()
            
            if product.find('span', attrs={'data-test':'product-leaf-piece-count-desktop-label'}):
                num_piece = product.find('span', attrs={'data-test':'product-leaf-piece-count-desktop-label'}).get_text()
            else :
                num_piece = product.find('span', attrs={'data-test':'product-leaf-piece-count-label'}).get_text()
            price = product.find('span', attrs={'data-test':'product-leaf-price'}).get_text()
            dis_price = product.find('span', attrs={'data-test':'product-leaf-discounted-price'})
            if dis_price != None:
                dis_price = product.find('span', attrs={'data-test':'product-leaf-discounted-price'}).get_text()
            else: dis_price = 'NoSale'
            print(name, age, num_piece, price, dis_price)
            
            # img = product.find("picture", attrs={'data-test':'product-leaf-image-1'})
            imgurl = product.find('picture').find('source').get('srcset').split('?')[0]
            print(product)
            print(imgurl)
            r = requests.get(imgurl, headers=hdr)
            with open(f'{cur_dir}/{img_dir_name}/{name.replace("/","_")}.jpg', "wb") as out:
                    out.write(r.content)
                
            prod_inf = [name, age, num_piece, price, dis_price]
            writer.writerow(prod_inf)
        
    

f.close()
```
