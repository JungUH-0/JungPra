```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
import time
import csv
```

```python
def page_down(num):
    body = driver.find_element(By.CSS_SELECTOR, 'body')
    #body.click()
    for i in range(num):
        body.send_keys(Keys.PAGE_DOWN)
        time.sleep(3)
```

```python
# https://help.naver.com/service/5626/contents/959?lang=ko
query = '한국폴리텍대학 "인공지능"'
url = 'https://www.naver.com/'
driver = webdriver.Chrome()

driver.get(url)
#time.sleep(3) # 안정적으로 웹사이트에 접근하기위해
driver.implicitly_wait(time_to_wait=10)
time.sleep(3) # 접속 후 잠시 멈춰서 봇인걸 알리지않기위해

search_box = driver.find_element(By.CSS_SELECTOR, "input#query")
search_box.send_keys(query)
search_box.send_keys(Keys.RETURN) #키를 누르는 코드
driver.implicitly_wait(time_to_wait=10) # 안정적으로 응답할때까지 대기
time.sleep(3)

#driver.find_element(By.LINK_TEXT, '다음').click()
while True:
    btn = driver.find_element(By.LINK_TEXT, '뉴스')
    if (btn.is_displayed):        
        btn.click()
        driver.implicitly_wait(time_to_wait=10)
        time.sleep(3)
        break        
    else:        
        driver.find_element(By.LINK_TEXT, '다음').click()
        
page_down(10)
```

```python
filename = "네이버뉴스검색정보_한국폴리텍대학_인공지능.csv"
f = open(filename, "w", encoding="utf-8-sig", newline="")
writer = csv.writer(f)

attributes = ["신문사", "일자", "제목", "주소", "내용"]
writer.writerow(attributes)

for i in range(len(news_press)):
    press = news_press[i].text.split("\n")[0]
    title = news_titles[i].text.split("\n")[0]
    des = news_des[i].text.split("\n")[0]
    date = news_date[i].text    
    rink = news_titles[i].get_attribute("href")    
    data_rows = [press, date, title, rink, des]
    writer.writerow(data_rows)        
    print(data_rows)
    
f.close()
driver.close()
```
