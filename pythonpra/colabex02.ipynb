````python
#같은 변수명을 끌고 와 쓸거면 global 사용
#같은 변수명을 사용시 들여쓰기가 많은 위치에있는 변수부터 사용(자기안에있는 변수부터 사용한다)
def outer_func():
    global y #전역 변수불러오기
    y=20 #전역 변수 의 값 변경
    def inner_func():
        global y
        global z #밖에서 선언안했지만 밖에서 선언한 취급
        z = 10
        y=30
        print('inner y=',y)
    inner_func()
    print('outer y=',y)
y=10 #전역 변수
outer_func()
print('global y=',y)
print('z=',z)
````

````python
def ex_func(int_a) :
    print("{}의 약수 : 1,".format(int_a), end=' ')

    for i in range(2,int_a):
        if int_a% i ==0:
            print("{},".format(i),end=" ")
    print(int_a)
````

````python
while True :
    num = int (input("약수 구할 정수 입력 :"))
    ex_func(num)

    rep = input("더이상 약수를 구하지 않을 거면 'n'을 입력 :")
    if rep == 'n':
        break
````

# Anonymous/Lambda function

• filter(), map()  등의 내장 함수에 사용

**lambda argments : expression**

double = lambda x : x*2


````python
my_list = [1,5,4,6,8,11,3,12]
new_list=[]
new_list = list(filter(lambda x: (x%2 == 0) , my_list))
print(new_list)
new_list=[]
new_list = list(map(lambda x : x*2 , my_list))
print(new_list)

def filter_1 (list):
    a= []
    for i in range(len(list)):
        # print(list[i])
        if list[i]%2==0:
            a.append(list[i])
    return a

def map_1(list) :
    a=[]
    for i in range(len(list)):
        a.append(list[i]*2)
    return a

new_list =[]
print(new_list)
new_list = map_1(my_list)
print(new_list)
# new_list = filter_1(my_list)
# print(new_list)
````
