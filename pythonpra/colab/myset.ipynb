```python
class data_Set :
    def __init__(self,size) :
        self.data_set = []
        self.maxsize=size
        self.length = 0

    def isFull(self) :
        if self.maxsize == self.length :
            return True
        else:
            return False

    def  isEmpty (self) :
        if self.length == 0 :
            return True
        else:
            return False

    def Contain(self,e) :
        if self.isEmpty() :
            print("Empty")
            return False
        else :
            for i in  self.data_set :
                if e == i :
                    return True
            return False

    def insert(self, e) :
        chk = self.Contain(e)
        chk1= self.isFull()
        if  chk1 == False :
            if chk == False :
                self.data_set += [e]
                self.length +=1
            else :
                print ("중복")
        else :
            print("Full")

    def delete(self, e):
        if self.isEmpty():
            print("Empty")
            return
        if self.Contain(e) == False:
            print("no data")
            return
        for i in range(self.length):
            if self.data_set[i] == e:
                for j in range(i, self.length-1):
                    self.data_set[j] = self.data_set[j+1]
                self.data_set[self.length-1] = None
                self.length -= 1
                return

    def Union(self,setB) :
        len1 = self.length
        len2 = setB.length
        setC = data_Set(len1+len2)
        for i in self.data_set :
            if i == None :
                continue
            setC.insert(i)
        for i in setB.data_set :
            if i == None :
                continue
            setC.insert(i)
        return setC

    def intersect(self,setB) :
        # if self.length < setB.length :
        #     len = self.length
        # else : len = setB.length
        min_len = self.length if self.length < setB.length else setB.length
        setC = data_Set(min_len)
        for i in setB.data_set :
            for j in self.data_set:
                if i == j :
                    setC.insert(i)
        return setC

    def difference(self,setB) :
        setC = data_Set(self.length)
        for i in self.data_set:
            if setB.Contain(i) :
                continue
            else :
                setC.insert(i)
        return setC
```
