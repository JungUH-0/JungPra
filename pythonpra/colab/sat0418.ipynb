```python
#문제 소수 판별
#인풋을 받을건지 랜덤으로 할건지
#1개의 숫자만인지 여러개인지




result = True
while result:
    count = 0
    try :
        p_num = int(input("숫자를 입력하세요"))
        if p_num != 0 and p_num>1:
            for i in range(1,p_num+1):
                if p_num%i == 0:
                    count +=1
                    # print(count)
            if count == 2 :
                print(f"{p_num}은 소수입니다.")
                result = False
            else :
                print(f"{p_num}은 소수가 아닙니다.")
        else :
            print("0,1,음수는 입력 불가합니다")

    except ValueError :
        print("숫자만 입력하세요")



```

```python

result = True
while result:
    count = 2 # count 값 초기화
    try :
        p_num = int(input("숫자를 입력하세요 : "))
        if p_num ==2 or p_num==3 : # 2/3 예외 발생
            print(f"{p_num}은 소수입니다.")
            result = False

        if p_num != 0 and p_num>1:
            for i in range(2,p_num-1):
                if p_num%i == 0:
                    count -=1
                if count != 2 : # count 값이 떨어지자마자 for문 종료시키기
                    print(f"{p_num}은 소수가 아닙니다.")
                    break

            if  count ==2:
                print(f"{p_num}은 소수입니다.")
                result = False
                break

        elif p_num<1 :
            print("0,1,음수는 입력 불가합니다")

    except ValueError :
        print("숫자만 입력하세요")
```

```python
# 비밀번호 검증
#1) 길이 8자
#2) 숫자,문자,특수(!,#,@)
#3) 대문자 1개 이상

# 입력받은 문자
# 입력받은 문자를 슬라이스하여 리스트1에 넣고 길이 확인
# 리스트에서 숫자와 문자를 구분하여 숫자가 있는지 확인 type
# 리스트1에서 숫자를 제외하고 리스트 2에 저장
# 리스트2에서 각 인덱스 돌려서 대문자 확인 여기서 문자가 없는지 확인 가능
# 리스트2에서 특수문자 확인
# 비밀번호 검증 완
#

result = None

while result == None or result ==False:
    pw_list = []
    pw_int = []
    pw_str = []
    pw_spstr=[]
    result = None
    print("비밀번호 숫자,문자,특수문자(!,#,@) 존재해야합니다.")
    pw= input("비밀번호 8자리 이상를 입력하세요 : ")
    for i in pw : # 입력받은 문자를 슬라이스해서 리스트 저장
        pw_list.append(i)
    for i in pw_list: # 저장한 리스트를 각 속성별로 찢어서 저장
        if ord(i)>=43 and ord(i)<=57:
            pw_int.append(i)
        elif (ord(i)>=65 and ord(i)<=90) or (ord(i)>=97 and ord(i)<=122 ):
            pw_str.append(i)
        elif ord(i)==33 or ord(i)==35 or ord(i)==64:
            pw_spstr.append(i)
        else:
            print("설정이 불가능한 문자가 존재합니다.재입력해주세요")
            result = False
            continue
    if len(pw_int) == 0:
        print("숫자가 없습니다. 숫자를 넣어주세요.")
        result = False
    if len(pw_str) ==0 :
        print("문자가가 없습니다. 문자를 넣어주세요.")
        result = False
    if len(pw_spstr) ==0 :
        print("특수문자가 없습니다. !,@,# 중 하나를 넣어주세요.")
        result = False

    spchk = 0;
    for i in pw_str :
        if ord(i)>=65 and ord(i)<=90:
            spchk+=1
    if spchk ==0 :
        print("대문자가 없습니다. 대문자를 넣어주세요.")
        result = False

    if len(pw_list)<8 :
        # print(len(pw_list))
        print("8자리 이상 설정해주세요.")
        result = False

    if result == None:
        print("비밀번호 설정 완료")
        result = True

```

```python
pw_list = []
pw_int = []
pw_str = []
pw_spstr=[]
agree_int="1234567890"#["1","2","3","4","5","6","7","8","9","0"]
agree_smallstr="abcdefghijklnmopqrstuwxyz"
agree_bigstr ="ABCDEFGHIJKLNMOPQRSTUWXYZ"
agree_str = [agree_smallstr,agree_bigstr]
agree_spstr = ["!","@","#"]
bigstr_chk =False
resultchk = True
result=False

print("비밀번호 숫자,문자,특수문자(!,#,@) 존재해야합니다.")
pw= input("비밀번호 8자리 이상를 입력하세요 : ")

for i in pw :
    pw_list.append(i)
    if i in agree_int:
        pw_int.append(i)
    elif i in agree_smallstr or i in agree_bigstr :
        pw_str.append(i)
        if i in agree_bigstr:
            bigstr_chk=True
    elif i in agree_spstr :
        pw_spstr.append(i)
    else:
        resultchk = False
        print("재입력해주세요.")

if resultchk :
    if len(pw_int)>0 and len(pw_str)>0 and len(pw_spstr)>0:
        if len(pw_list)>7:
            if bigstr_chk :
                result = True

print(result)
```
