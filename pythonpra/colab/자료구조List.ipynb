```python
class data_List :
    def __init__(self,size):
        self.data_list=[None]*size
        self.size=size
        self.length=0

    def make_IsFull (self):
        if self.size == self.length :
            return True
        else:
            return False

    def  make_IsEmpty (self):
        if self.length == 0 :
            return True
        else:
            return False
    def make_Print(self):
        pass


    def make_Insert (self,pos,e):

        if self.make_IsFull() == True :
            return print("size over")

        if self.data_list[pos] != None :
            temp_ori = self.data_list[pos] #원본데이터
            self.data_list[pos]=e
            for i in range(pos, self.length-1):
                temp = self.data_list[i+1]
                self.data_list[i+1] = temp_ori
                temp_ori = temp

            self.data_list[self.length] = temp_ori # 마지막값
            self.length +=1
        else :
            self.data_list[pos] = e
            self.length +=1

    def make_Delete (self,pos):
        if self.make_IsEmpty() == True :
            return print("empty")
        if self.data_list[pos] == None :
            return print("already None")
        else :
            temp_ori = self.data_list[self.length-1]
            print(temp_ori)
            for i in range(self.length-2,pos-1,-1):
                print (i)
                temp = self.data_list[i]
                print (temp)
                self.data_list[i] = temp_ori
                temp_ori=temp
            self.data_list[self.length-1] = None
            self.length -=1

    def make_getEntry(self,pos):
        if self.data_list[pos] == None :
            return print("None")
        else :
            return self.data_list[pos]

    def make_Size(self):
        return self.length

    def make_Clear(self):
        for i in range(self.length):
            self.data_list[i] = None

    def make_Find(self,item):
        result = None
        for i in range(self.length):

            if self.data_list[i] == item :
                print("item의 위치는",item,i)
                result = i
        if result != None :
            return result
        else :
            return print("no data")

    def make_Replace(self,pos,item) :
        self.data_list[pos] = item

    def make_Append(self,e):

        self.data_list[self.length] = e
        self.length += 1



my_list = data_List(10)
my_list.make_Delete(8)
print("length",my_list.length)
my_list.make_Insert(0,100)
my_list.make_Insert(1,10)
my_list.make_Insert(2,20)
my_list.make_Insert(3,30)
my_list.make_Insert(4,40)
my_list.make_Insert(5,50)
my_list.make_Insert(6,60)
my_list.make_Insert(7,70)
my_list.make_Insert(8,80)
print("length",my_list.length)
print(my_list.data_list)
# print("empty",my_list.make_IsEmpty())
# print("full",my_list.make_IsFull())
my_list.make_Insert(9,90)
print(my_list.data_list)
my_list.make_Delete(0)
print(my_list.data_list)
my_list.make_Insert(0,8888)
my_list.make_Delete(3)
my_list.make_Delete(4)
# print("length",my_list.length)
print(my_list.data_list)
print("length",my_list.length)
my_list.make_Delete(8)
print(my_list.make_getEntry(0))
for i in range(my_list.length):
    print(my_list.make_getEntry(i))

print(my_list.make_Find(100))
# my_list.make_Clear()
print(my_list.data_list)
my_list.make_Append(9999)
print(my_list.data_list)
```
