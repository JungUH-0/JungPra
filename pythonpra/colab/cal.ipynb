```python

# 예외처리 함수

def chk(a)  :
    re_list=[]
    for i in a :
        if type(i) == int :
            re_list.append(i)# 새 리스트에 저장
        else :
            continue
    return re_list

list_t = [0,1,2,3,4,5,'a',-6,'b',10,'t',0]
print(list_t)
list_t= chk(list_t)
print(list_t)
print(len(list_t))
```

```python
#계산 함수 (4개)
def plus (a):
    result = 0
    for i in a :
        result += i
    return result



def minus (a):
    result = 0
    for i in range(len(a)) :
        if result ==0 and i==0 :
            result = a[i]
        else :
            result -= a[i]
    return result
print(minus([1,1,1,1,1,1,1]))

def multip (a):
    result = 0
    for i in range(len(a)) :
        if result ==0 :
            result = a[i]
        else :
            result *= a[i]
    return result


def divi (a):
    result = 0
    for i in range(len(a)) :
        if(a[i]) != 0:
            if result == 0 : # 인덱스 0 이 숫자 0 이면? 해결방법은?
                result = a[i]
            else :
                result /= a[i]
        else :
            print (i)
            pass
    return result


# 평균
def avg(a):
    sum=plus(a)
    return int(sum/len(a))


#최대
def maxnum(a):
    maxnum = 0
    for i in a :
        if maxnum< i:
            maxnum = i
    return maxnum
    # return max(a)


#최소
def mini(a):
    minum = 0
    for i in a :
        if minum> i:
            minum = i
    return minum
    # return min(a)
```
